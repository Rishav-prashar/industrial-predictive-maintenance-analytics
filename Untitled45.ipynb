```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

# 1st kpi ( Failure Percentage)

```python
failure_rate = data['Machine failure'].mean() * 100

print("Machine Failure Rate:", failure_rate, "%")
```

```python
sns.countplot(x='Machine failure', data=data)

plt.title('Machine Failure Distribution')
plt.xlabel('Machine Failure')
plt.ylabel('Number of Observations')

plt.show()
```

# failure percentage for each product type

```python
data.groupby('Type')['Machine failure'].mean() * 100
```

```python
failure_by_type = data.groupby('Type')['Machine failure'].mean() * 100

failure_by_type.plot(kind='bar')

plt.title('Machine Failure Rate by Product Type')
plt.xlabel('Product Type')
plt.ylabel('Failure Rate (%)')
plt.xticks(rotation=0)

plt.show()
```

# torque vs machine failure

```python
data.groupby('Machine failure')['Torque [Nm]'].mean()
```

```python
sns.boxplot(x='Machine failure', y='Torque [Nm]', data=data)

plt.title('Torque vs Machine Failure')
plt.xlabel('Machine Failure (0 = No, 1 = Yes)')
plt.ylabel('Torque [Nm]')

plt.show()
```

# compare torque for each operating mode

```python
failure_modes = ['TWF', 'HDF', 'PWF', 'OSF', 'RNF']

for mode in failure_modes:
    print(mode)
    print(data.groupby(mode)['Torque [Nm]'].mean())
    print()
```

# toolwear

```python
for mode in failure_modes:
    print(mode)
    print(data.groupby(mode)['Tool wear [min]'].mean())
    print()
```

```python
failure_modes = ['TWF', 'HDF', 'PWF', 'OSF', 'RNF']

data['Number of Failure Modes'] = data[failure_modes].sum(axis=1)

data['Number of Failure Modes'].value_counts().sort_index()
```

```python
pd.crosstab(
    data['Machine failure'],
    data['Number of Failure Modes']
)
```

# corelation matrix

```python
corr = data[
    ['Air temperature [K]',
     'Process temperature [K]',
     'Rotational speed [rpm]',
     'Torque [Nm]',
     'Tool wear [min]',
     'Machine failure']
].corr()

corr
```

# KPI SECTION

```python
total_observations = len(data)

total_failures = data['Machine failure'].sum()

failure_rate = data['Machine failure'].mean() * 100

average_torque = data['Torque [Nm]'].mean()

average_tool_wear = data['Tool wear [min]'].mean()

print("Total observations:", total_observations)
print("Total machine failures:", total_failures)
print("Machine failure rate:", round(failure_rate, 2), "%")
print("Average torque:", round(average_torque, 2), "Nm")
print("Average tool wear:", round(average_tool_wear, 2), "min")
```
